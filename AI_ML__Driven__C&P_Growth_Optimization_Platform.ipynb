```python
# ============================================================
# AHADU BANK
# AI DRIVEN CUSTOMER & PRODUCT GROWTH OPTIMIZATION PLATFORM
# ============================================================

import warnings
warnings.filterwarnings("ignore")
# ── Install Gradio (For Demo) ──────────────────
!pip install gradio -q
print("✅ Gradio installed")
import os, re, json, random, logging
import numpy as np
import pandas as pd
# ── Colab-compatible matplotlib (NO Agg backend) ─────────────
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams["figure.dpi"] = 100
import seaborn as sns

from datetime import datetime
from pathlib import Path

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, mean_squared_error, mean_absolute_error,
    silhouette_score, davies_bouldin_score,
    precision_score, recall_score, f1_score
)
from sklearn.decomposition import PCA
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier

try:
    import xgboost as xgb;  XGBOOST_AVAILABLE = True;  print("✅ XGBoost ready")
except ImportError:
    XGBOOST_AVAILABLE = False;  print("⚠️  XGBoost not found — will use GradientBoosting")

try:
    import lightgbm as lgb;  LGBM_AVAILABLE = True;  print("✅ LightGBM ready")
except ImportError:
    LGBM_AVAILABLE = False;  print("⚠️  LightGBM not found — will use RandomForest")

try:
    import shap;  SHAP_AVAILABLE = True;  print("✅ SHAP ready")
except ImportError:
    SHAP_AVAILABLE = False;  print("⚠️  SHAP not found — explainability will be skipped")

# ── Config ────────────────────────────────────────────────────
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

OUTPUT_DIR = Path("outputs_real")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ── Colab-compatible logging (prints to notebook AND file) ────
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    handlers=[
        logging.StreamHandler(),                                    # shows in Colab cell output
        logging.FileHandler(OUTPUT_DIR / "pipeline_real.log"),     # also saves to file
    ]
)
log = logging.getLogger(__name__)

PRODUCT_LIST = [
    "Ordinary Saving Account",    "Blatena Saving Account",
    "Werazut Saving Account",     "Dekemezmur Saving Account",
    "Ahati Saving Account",       "Abew Saving Account",
    "Special Saving Account",     "Hiruyan Saving Account",
    "Tirufat Savings Account",    "Asrat Bekurat Saving Account",
    "Ahadu Bereket Saving Account","Janderebaw Saving Account",
    "Prime Customer Deposit Account","Equb Saving Account",
    "Agape Saving Account",       "Couples Saving Deposit",
    "Washera Saving Account",     "Sitota Saving Deposit",
]

print("\n✅ Cell 1 done — imports and config loaded")


# ─────────────────────────────────────────────────────────────
# CELL 2 — Load Real Data
# ─────────────────────────────────────────────────────────────
def load_real_data(filepath: str) -> pd.DataFrame:
    print(f"\n📂 Loading: {filepath}")
    ext = Path(filepath).suffix.lower()

    if ext == ".csv":
        df = pd.read_csv(filepath)
    elif ext in (".xlsx", ".xls"):
        df = pd.read_excel(filepath)
    else:
        raise ValueError(f"Unsupported file type: {ext}")

    # Parse date columns safely
    for date_col in ["Opening Date", "Last_Transaction_Date"]:
        if date_col in df.columns:
            df[date_col] = pd.to_datetime(df[date_col], errors="coerce", dayfirst=False)

    # Enforce integer columns
    int_cols = ["Age", "ACTIVE_ACCOUNT", "INOPERATIVE", "DORMANT",
                "Debit Freez", "Credit Freez", "Days_Since_Last_Txn", "RFM_Score"]
    for col in int_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)

    # Enforce float columns
    float_cols = ["Initial_Deposit", "End Balance", "Account_Tenure_Years",
                  "CLV_Predicted", "Churn_Probability", "Dormancy_Risk_Score", "Lookalike_Score"]
    for col in float_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0.0)

    print(f"✅ Loaded {df.shape[0]:,} rows × {df.shape[1]} columns")
    print(f"   Columns: {df.columns.tolist()}")
    return df

# ── Run it ────────────────────────────────────────────────────
# Change the path below to match where you uploaded your file
df = load_real_data("/content/ahadu_bank_enriched_real.csv")
df.head(3)


# ─────────────────────────────────────────────────────────────
# CELL 3 — Feature Engineering
# ─────────────────────────────────────────────────────────────
def feature_engineering(df: pd.DataFrame) -> pd.DataFrame:
    print("\n⚙️  Running feature engineering …")
    df = df.copy()
    today = pd.Timestamp("2025-12-31")

    df["Opening Date"]          = pd.to_datetime(df["Opening Date"],          errors="coerce")
    df["Last_Transaction_Date"] = pd.to_datetime(df["Last_Transaction_Date"], errors="coerce")

    df["Account_Tenure_Days"]  = (today - df["Opening Date"]).dt.days.fillna(0).astype(int)
    df["Account_Tenure_Years"] = (df["Account_Tenure_Days"] / 365.25).round(2)
    df["Days_Since_Last_Txn"]  = (today - df["Last_Transaction_Date"]).dt.days.fillna(9999).astype(int)
    df["Opening_Year"]         = df["Opening Date"].dt.year
    df["Opening_Month"]        = df["Opening Date"].dt.month

    df["Balance_Growth"]       = df["End Balance"] - df["Initial_Deposit"]
    df["Balance_Growth_Rate"]  = (df["Balance_Growth"] / (df["Initial_Deposit"] + 1)).replace([np.inf, -np.inf], 0).round(4)
    df["Log_End_Balance"]      = np.log1p(df["End Balance"])
    df["Log_Initial_Deposit"]  = np.log1p(df["Initial_Deposit"])
    df["Balance_to_Deposit"]   = (df["End Balance"] / (df["Initial_Deposit"] + 1)).replace([np.inf, -np.inf], 0).round(4)

    df["RFM_Recency"]   = df["Days_Since_Last_Txn"]
    df["RFM_Frequency"] = (1 / (df["Days_Since_Last_Txn"] + 1)).round(6)
    df["RFM_Monetary"]  = df["End Balance"]

    df["R_Score"] = pd.qcut(df["RFM_Recency"].rank(method="first"),   q=5, labels=[5,4,3,2,1]).astype(int)
    df["F_Score"] = pd.qcut(df["RFM_Frequency"].rank(method="first"), q=5, labels=[1,2,3,4,5]).astype(int)
    df["M_Score"] = pd.qcut(df["RFM_Monetary"].rank(method="first"),  q=5, labels=[1,2,3,4,5]).astype(int)
    df["RFM_Score"]   = df["R_Score"] + df["F_Score"] + df["M_Score"]
    df["RFM_Segment"] = pd.cut(df["RFM_Score"], bins=[2,5,8,11,15],
                                labels=["Low Value","Mid Value","High Value","Champions"])

    df["Sex_Encoded"]     = (df["Sex"] == "M").astype(int)
    le_prod               = LabelEncoder()
    df["Product_Encoded"] = le_prod.fit_transform(df["Product Type"].fillna("Unknown"))
    le_reg                = LabelEncoder()
    df["Region_Encoded"]  = le_reg.fit_transform(df["Region"].fillna("Unknown"))

    df["Age_Group"] = pd.cut(df["Age"], bins=[17,25,35,45,55,75],
                              labels=["18-25","26-35","36-45","46-55","56+"])

    def account_status(row):
        if row["DORMANT"]     == 1: return 2
        if row["INOPERATIVE"] == 1: return 1
        if row["Debit Freez"] == 1 or row["Credit Freez"] == 1: return 3
        return 0
    df["Account_Status_Code"] = df.apply(account_status, axis=1)

    if "CLV_Predicted" in df.columns:
        df["CLV_Target"] = df["CLV_Predicted"].fillna(0)
        print("   CLV_Target ← real CLV_Predicted column from file")
    else:
        clv_base = (df["End Balance"]*0.4 + df["Initial_Deposit"]*0.1
                    + df["Account_Tenure_Years"]*5000 + df["RFM_Score"]*2000
                    - df["Days_Since_Last_Txn"]*10)
        df["CLV_Target"] = np.clip(clv_base, 0, None).round(2)
        print("   CLV_Target ← computed from formula (no CLV column found)")

    df["Product_Count"] = df.groupby("CIF")["Product Type"].transform("count")

    print(f"✅ Feature engineering done — shape: {df.shape}")
    return df

df = feature_engineering(df)
df.head(3)


# ─────────────────────────────────────────────────────────────
# CELL 4 — EDA Dashboard
# ─────────────────────────────────────────────────────────────
def run_eda(df: pd.DataFrame):
    print("\n📊 Running EDA …")
    fig, axes = plt.subplots(3, 4, figsize=(24, 18))
    fig.suptitle("AHADU BANK AI DRIVEN CUSTOMER & PRODUCT GROWTH OPTIMIZATION PLATFORM DASHBOARD ", fontsize=16, fontweight="bold")

    axes[0,0].hist(df["Age"], bins=30, color="#1f77b4", edgecolor="white")
    axes[0,0].set_title("Age Distribution"); axes[0,0].set_xlabel("Age")

    sx = df["Sex"].value_counts()
    axes[0,1].pie(sx, labels=sx.index, autopct="%1.1f%%", colors=["#4C72B0","#DD8452"])
    axes[0,1].set_title("Gender Distribution")

    status_map = {0:"Active",1:"Inoperative",2:"Dormant",3:"Frozen"}
    sc = df["Account_Status_Code"].map(status_map).value_counts()
    axes[0,2].bar(sc.index, sc.values, color=["#2ca02c","#ff7f0e","#d62728","#9467bd"])
    axes[0,2].set_title("Account Status Breakdown")

    tp = df["Product Type"].value_counts().head(10)
    axes[0,3].barh(tp.index[::-1], tp.values[::-1], color="#17becf")
    axes[0,3].set_title("Top 10 Products")

    axes[1,0].hist(np.log1p(df["End Balance"]), bins=40, color="#e377c2", edgecolor="white")
    axes[1,0].set_title("Log(End Balance)")

    axes[1,1].hist(df["Account_Tenure_Years"], bins=30, color="#bcbd22", edgecolor="white")
    axes[1,1].set_title("Account Tenure (Years)")

    axes[1,2].hist(df["Days_Since_Last_Txn"].clip(0,1000), bins=40, color="#8c564b", edgecolor="white")
    axes[1,2].set_title("Days Since Last Txn (capped 1000)")

    rfm_c = df["RFM_Segment"].value_counts()
    axes[1,3].bar(rfm_c.index.astype(str), rfm_c.values,
                  color=["#d62728","#ff7f0e","#2ca02c","#1f77b4"])
    axes[1,3].set_title("RFM Segment Distribution")

    rc = df["Region"].value_counts()
    axes[2,0].barh(rc.index[::-1], rc.values[::-1], color="#7f7f7f")
    axes[2,0].set_title("Customers by Region")

    ab = df.groupby("Age_Group", observed=True)["End Balance"].median()
    axes[2,1].bar(ab.index.astype(str), ab.values, color="#1f77b4")
    axes[2,1].set_title("Median Balance by Age Group")

    num_cols = ["Age","Initial_Deposit","End Balance","Account_Tenure_Years",
                "Days_Since_Last_Txn","RFM_Score","Balance_Growth_Rate"]
    corr = df[num_cols].corr()
    im = axes[2,2].imshow(corr, cmap="RdYlGn", vmin=-1, vmax=1)
    axes[2,2].set_xticks(range(len(num_cols)))
    axes[2,2].set_yticks(range(len(num_cols)))
    axes[2,2].set_xticklabels([c[:8] for c in num_cols], rotation=45, ha="right")
    axes[2,2].set_yticklabels([c[:8] for c in num_cols])
    axes[2,2].set_title("Correlation Heatmap")
    plt.colorbar(im, ax=axes[2,2])

    axes[2,3].hist(np.log1p(df["CLV_Target"]), bins=40, color="#aec7e8", edgecolor="white")
    axes[2,3].set_title("Log(CLV Target)")

    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "eda_dashboard_real.png", dpi=120, bbox_inches="tight")
    plt.show()   # ← this line makes it visible in Colab
    print(f"✅ EDA chart shown and saved to outputs_real/eda_dashboard_real.png")

    print("\n" + "="*55)
    print(" DATA SUMMARY")
    print("="*55)
    print(f"  Total customers      : {len(df):,}")
    print(f"  Active accounts      : {df['ACTIVE_ACCOUNT'].sum():,}  ({df['ACTIVE_ACCOUNT'].mean()*100:.1f}%)")
    print(f"  Inoperative accounts : {df['INOPERATIVE'].sum():,}  ({df['INOPERATIVE'].mean()*100:.1f}%)")
    print(f"  Dormant accounts     : {df['DORMANT'].sum():,}  ({df['DORMANT'].mean()*100:.1f}%)")
    print(f"  Avg End Balance      : ETB {df['End Balance'].mean():,.2f}")
    print(f"  Median End Balance   : ETB {df['End Balance'].median():,.2f}")
    print(f"  Avg Account Tenure   : {df['Account_Tenure_Years'].mean():.1f} years")
    print("="*55)

run_eda(df)


# ─────────────────────────────────────────────────────────────
# CELL 5 — Customer Segmentation
# ─────────────────────────────────────────────────────────────
SEG_FEATURES = [
    "Age","Log_End_Balance","Account_Tenure_Years",
    "Days_Since_Last_Txn","RFM_Score","Balance_Growth_Rate",
    "Sex_Encoded","Region_Encoded","Product_Encoded",
]

def build_segmentation(df: pd.DataFrame) -> pd.DataFrame:
    print("\n🔵 Training customer segmentation …")
    X        = df[SEG_FEATURES].fillna(0)
    scaler   = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Elbow curve
    inertias = []
    for k in range(2, 12):
        km = KMeans(n_clusters=k, random_state=RANDOM_SEED, n_init=10)
        km.fit(X_scaled)
        inertias.append(km.inertia_)

    plt.figure(figsize=(8,4))
    plt.plot(range(2,12), inertias, "bo-")
    plt.xlabel("Number of Clusters (k)"); plt.ylabel("Inertia")
    plt.title("K-Means Elbow Curve ")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "elbow_curve_real.png", dpi=100)
    plt.show()   # ← visible in Colab

    N_CLUSTERS = 6
    kmeans = KMeans(n_clusters=N_CLUSTERS, random_state=RANDOM_SEED, n_init=15)
    df["Segment_KMeans"] = kmeans.fit_predict(X_scaled)

    sil = silhouette_score(X_scaled, df["Segment_KMeans"])
    dbi = davies_bouldin_score(X_scaled, df["Segment_KMeans"])
    print(f"   K-Means Silhouette: {sil:.4f}  |  DB-Index: {dbi:.4f}")

    pca   = PCA(n_components=2, random_state=RANDOM_SEED)
    X_pca = pca.fit_transform(X_scaled)
    palette = sns.color_palette("tab10", N_CLUSTERS)
    plt.figure(figsize=(10,7))
    for seg in range(N_CLUSTERS):
        mask = df["Segment_KMeans"] == seg
        plt.scatter(X_pca[mask,0], X_pca[mask,1],
                    label=f"Segment {seg}", alpha=0.4, s=8, color=palette[seg])
    plt.legend(markerscale=3, fontsize=9)
    plt.title(f"Customer Segmentation PCA — Real Data  (Silhouette={sil:.3f})")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "segmentation_pca_real.png", dpi=120)
    plt.show()   # ← visible in Colab

    seg_profiles = df.groupby("Segment_KMeans").agg(
        Avg_Balance=("End Balance","mean")
    ).sort_values("Avg_Balance", ascending=False)
    labels = ["Champions","High-CLV Active","Mid-Value Growth",
              "Cross-Sell Ready","Dormant Risk","Low Engagement"]
    label_map = {row.Index: labels[min(i, len(labels)-1)]
                 for i, row in enumerate(seg_profiles.itertuples())}
    df["Segment_Label"] = df["Segment_KMeans"].map(label_map)

    dbscan = DBSCAN(eps=0.8, min_samples=10, n_jobs=-1)
    df["DBSCAN_Label"] = dbscan.fit_predict(X_scaled)
    print(f"   DBSCAN noise points: {(df['DBSCAN_Label']==-1).sum():,}")

    seg_summary = df.groupby("Segment_Label").agg(
        Count          =("CIF","count"),
        Avg_Balance    =("End Balance","mean"),
        Avg_CLV        =("CLV_Target","mean"),
        Avg_RFM        =("RFM_Score","mean"),
        Dormant_Rate   =("DORMANT","mean"),
        Avg_Tenure_Yrs =("Account_Tenure_Years","mean"),
    ).round(2)
    seg_summary.to_csv(OUTPUT_DIR / "segment_summary_real.csv")
    print(f"\n✅ Segment Summary:\n")
    display(seg_summary)   # ← pretty table in Colab
    return df

df = build_segmentation(df)


# ─────────────────────────────────────────────────────────────
# CELL 6 — CLV Prediction
# ─────────────────────────────────────────────────────────────
CLV_FEATURES = [
    "Age","Sex_Encoded","Region_Encoded","Product_Encoded",
    "Log_Initial_Deposit","Log_End_Balance","Account_Tenure_Years",
    "Days_Since_Last_Txn","RFM_Score","R_Score","F_Score","M_Score",
    "Balance_Growth_Rate","Balance_to_Deposit","Product_Count",
    "Account_Tenure_Days","DORMANT","INOPERATIVE","ACTIVE_ACCOUNT",
    "Debit Freez","Credit Freez",
]

def train_clv_model(df: pd.DataFrame):
    print("\n💰 Training CLV prediction model …")
    X = df[CLV_FEATURES].fillna(0)
    y = df["CLV_Target"]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=RANDOM_SEED
    )

    if XGBOOST_AVAILABLE:
        model = xgb.XGBRegressor(
            n_estimators=500, max_depth=6, learning_rate=0.05,
            subsample=0.8, colsample_bytree=0.8,
            reg_alpha=0.1, reg_lambda=1.0,
            random_state=RANDOM_SEED, n_jobs=-1, verbosity=0
        )
    else:
        from sklearn.ensemble import GradientBoostingRegressor
        model = GradientBoostingRegressor(
            n_estimators=300, max_depth=5, learning_rate=0.05,
            random_state=RANDOM_SEED
        )

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    rmse   = np.sqrt(mean_squared_error(y_test, y_pred))
    mae    = mean_absolute_error(y_test, y_pred)
    mape   = np.mean(np.abs((y_test - y_pred) / (y_test + 1))) * 100

    print(f"   RMSE : {rmse:,.2f}")
    print(f"   MAE  : {mae:,.2f}")
    print(f"   MAPE : {mape:.2f}%")

    df["CLV_Predicted"] = model.predict(X)
    df["CLV_Tier"] = pd.qcut(df["CLV_Predicted"], q=4,
                              labels=["Low-CLV","Mid-CLV","High-CLV","Top-CLV"])

    # Feature importance chart
    if XGBOOST_AVAILABLE:
        imp = pd.Series(model.feature_importances_, index=CLV_FEATURES).sort_values()
        fig, ax = plt.subplots(figsize=(10,8))
        imp.tail(15).plot(kind="barh", ax=ax, color="#1f77b4")
        ax.set_title("CLV Feature Importances ")
        plt.tight_layout()
        plt.savefig(OUTPUT_DIR / "clv_feature_importance_real.png", dpi=120)
        plt.show()   # ← visible in Colab

    # SHAP
    if SHAP_AVAILABLE and XGBOOST_AVAILABLE:
        print("   Computing SHAP values …")
        explainer = shap.TreeExplainer(model)
        shap_vals = explainer.shap_values(X_test.iloc[:500])
        plt.figure()
        shap.summary_plot(shap_vals, X_test.iloc[:500],
                          feature_names=CLV_FEATURES, show=False)
        plt.tight_layout()
        plt.savefig(OUTPUT_DIR / "clv_shap_real.png", dpi=120, bbox_inches="tight")
        plt.show()   # ← visible in Colab

    print("✅ CLV model done")
    return model, df, {"rmse": rmse, "mae": mae, "mape": mape}

clv_model, df, clv_metrics = train_clv_model(df)


# ─────────────────────────────────────────────────────────────
# CELL 7 — Churn Prediction
# ─────────────────────────────────────────────────────────────
CHURN_FEATURES = [
    "Age","Sex_Encoded","Region_Encoded","Product_Encoded",
    "Log_End_Balance","Log_Initial_Deposit","Account_Tenure_Years",
    "Days_Since_Last_Txn","RFM_Score","R_Score","F_Score","M_Score",
    "Balance_Growth_Rate","Balance_to_Deposit","CLV_Predicted",
    "Debit Freez","Credit Freez","Product_Count",
]

def train_churn_model(df: pd.DataFrame):
    print("\n⚠️  Training churn prediction model …")
    X = df[CHURN_FEATURES].fillna(0)
    y = df["DORMANT"]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=RANDOM_SEED
    )
    pos_w = (y_train == 0).sum() / max((y_train == 1).sum(), 1)

    if LGBM_AVAILABLE:
        model = lgb.LGBMClassifier(
            n_estimators=500, max_depth=7, learning_rate=0.05,
            num_leaves=63, subsample=0.8, colsample_bytree=0.8,
            scale_pos_weight=pos_w, random_state=RANDOM_SEED,
            n_jobs=-1, verbose=-1
        )
    else:
        model = RandomForestClassifier(
            n_estimators=300, max_depth=8, class_weight="balanced",
            random_state=RANDOM_SEED, n_jobs=-1
        )

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    auc    = roc_auc_score(y_test, y_prob)
    f1     = f1_score(y_test, y_pred, zero_division=0)
    prec   = precision_score(y_test, y_pred, zero_division=0)
    rec    = recall_score(y_test, y_pred, zero_division=0)

    print(f"   AUC-ROC   : {auc:.4f}")
    print(f"   F1 Score  : {f1:.4f}")
    print(f"   Precision : {prec:.4f}")
    print(f"   Recall    : {rec:.4f}")
    print(f"\n{classification_report(y_test, y_pred, target_names=['Retained','Dormant'])}")

    df["Churn_Probability"] = model.predict_proba(X)[:, 1]
    df["Churn_Risk_Band"]   = pd.cut(df["Churn_Probability"], bins=[0,0.3,0.5,0.7,1.0],
                                      labels=["Low Risk","Medium Risk","High Risk","Critical Risk"])

    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred)
    fig, ax = plt.subplots(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=["Retained","Dormant"],
                yticklabels=["Retained","Dormant"], ax=ax)
    ax.set_title(f"Churn Confusion Matrix (AUC={auc:.4f})")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "churn_confusion_matrix_real.png", dpi=120)
    plt.show()   # ← visible in Colab

    # SHAP
    if SHAP_AVAILABLE and LGBM_AVAILABLE:
        print("   Computing SHAP values …")
        explainer = shap.TreeExplainer(model)
        sv = explainer.shap_values(X_test.iloc[:500])
        sv = sv[1] if isinstance(sv, list) else sv
        plt.figure()
        shap.summary_plot(sv, X_test.iloc[:500],
                          feature_names=CHURN_FEATURES, show=False)
        plt.tight_layout()
        plt.savefig(OUTPUT_DIR / "churn_shap_real.png", dpi=120, bbox_inches="tight")
        plt.show()   # ← visible in Colab

    print("✅ Churn model done")
    return model, df, {"auc_roc": auc, "f1": f1, "precision": prec, "recall": rec}

churn_model, df, churn_metrics = train_churn_model(df)


# ─────────────────────────────────────────────────────────────
# CELL 8 — Dormancy Detection
# ─────────────────────────────────────────────────────────────
DORMANCY_FEATURES = [
    "Age","Sex_Encoded","Region_Encoded","Product_Encoded",
    "Log_End_Balance","Account_Tenure_Years","Days_Since_Last_Txn",
    "Balance_Growth_Rate","RFM_Score","Churn_Probability",
    "Debit Freez","Credit Freez",
]

def train_dormancy_model(df: pd.DataFrame):
    print("\n😴 Training dormancy detection model …")
    X = df[DORMANCY_FEATURES].fillna(0)
    y = df["INOPERATIVE"]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=RANDOM_SEED
    )
    model = GradientBoostingClassifier(
        n_estimators=300, max_depth=5, learning_rate=0.05,
        subsample=0.8, random_state=RANDOM_SEED
    )
    model.fit(X_train, y_train)
    y_prob = model.predict_proba(X_test)[:, 1]
    y_pred = model.predict(X_test)
    auc    = roc_auc_score(y_test, y_prob)
    f1     = f1_score(y_test, y_pred, zero_division=0)
    print(f"   AUC-ROC : {auc:.4f}  |  F1: {f1:.4f}")

    df["Dormancy_Risk_Score"] = model.predict_proba(X)[:, 1]
    df["Dormancy_Risk_Band"]  = pd.cut(df["Dormancy_Risk_Score"], bins=[0,0.3,0.6,1.0],
                                        labels=["Safe","At Risk","High Dormancy Risk"])

    at_risk = df[
        (df["Dormancy_Risk_Score"] > 0.5) & (df["DORMANT"] == 0)
    ][["CIF","Account Number","Account Name","Product Type","Region",
       "End Balance","Days_Since_Last_Txn","Dormancy_Risk_Score"]
    ].sort_values("Dormancy_Risk_Score", ascending=False)

    at_risk.to_csv(OUTPUT_DIR / "dormancy_at_risk_real.csv", index=False)
    print(f"✅ Dormancy model done — {len(at_risk):,} at-risk accounts saved")
    display(at_risk.head(10))   # ← show top 10 in Colab
    return model, df

dormancy_model, df = train_dormancy_model(df)


# ─────────────────────────────────────────────────────────────
# CELL 9 — NBP Recommendations (all customers)
# ─────────────────────────────────────────────────────────────
def build_nbp_recommendations(df: pd.DataFrame) -> pd.DataFrame:
    print(f"\n🎯 Building NBP recommendations for {len(df):,} customers …")
    print("   (this may take 1-2 minutes for 10,000 customers)")

    seg_product = pd.crosstab(df["Segment_Label"], df["Product Type"], normalize="index")
    clv_product = pd.crosstab(df["CLV_Tier"],      df["Product Type"], normalize="index")
    age_product = pd.crosstab(df["Age_Group"],     df["Product Type"], normalize="index")

    nbp_records = []
    for _, row in df.iterrows():
        current = row["Product Type"]
        scores  = {}
        for prod in PRODUCT_LIST:
            if prod == current:
                continue
            score = 0.0
            seg = row.get("Segment_Label", "")
            if seg in seg_product.index and prod in seg_product.columns:
                score += seg_product.loc[seg, prod] * 0.40
            clv_t = str(row.get("CLV_Tier", ""))
            if clv_t in clv_product.index and prod in clv_product.columns:
                score += clv_product.loc[clv_t, prod] * 0.35
            age_g = str(row.get("Age_Group", ""))
            if age_g in age_product.index and prod in age_product.columns:
                score += age_product.loc[age_g, prod] * 0.25
            scores[prod] = round(score, 5)

        top3 = sorted(scores.items(), key=lambda x: x[1], reverse=True)[:3]
        for rank, (prod, conf) in enumerate(top3, 1):
            nbp_records.append({
                "CIF"                : row["CIF"],
                "Account_Name"       : row["Account Name"],
                "Current_Product"    : current,
                "Segment"            : row.get("Segment_Label",""),
                "CLV_Tier"           : str(row.get("CLV_Tier","")),
                "Churn_Risk"         : str(row.get("Churn_Risk_Band","")),
                "NBP_Rank"           : rank,
                "Recommended_Product": prod,
                "Confidence_Score"   : conf,
            })

    nbp_df = pd.DataFrame(nbp_records)
    nbp_df.to_csv(OUTPUT_DIR / "nbp_recommendations_real.csv", index=False)
    print(f"✅ NBP done — saved to outputs_real/nbp_recommendations_real.csv")
    display(nbp_df.head(10))   # ← show top 10 in Colab
    return nbp_df

nbp_df = build_nbp_recommendations(df)


# ─────────────────────────────────────────────────────────────
# CELL 10 — Cross-sell Propensity
# ─────────────────────────────────────────────────────────────
def train_crosssell_model(df: pd.DataFrame) -> pd.DataFrame:
    print("\n🔄 Training cross-sell propensity models …")
    XSELL_FEATURES = [
        "Age","Sex_Encoded","Region_Encoded",
        "Log_End_Balance","Account_Tenure_Years","RFM_Score",
        "CLV_Predicted","Churn_Probability","Balance_Growth_Rate",
    ]
    top_products = df["Product Type"].value_counts().head(5).index.tolist()
    for prod in top_products:
        col = "PropScore_" + re.sub(r"[^a-zA-Z0-9]","_",prod)[:30]
        y   = (df["Product Type"] == prod).astype(int)
        X   = df[XSELL_FEATURES].fillna(0)
        X_tr, X_te, y_tr, y_te = train_test_split(
            X, y, test_size=0.2, stratify=y, random_state=RANDOM_SEED
        )
        clf = GradientBoostingClassifier(
            n_estimators=150, max_depth=4, learning_rate=0.1,
            random_state=RANDOM_SEED
        )
        clf.fit(X_tr, y_tr)
        auc = roc_auc_score(y_te, clf.predict_proba(X_te)[:,1])
        print(f"   [{prod[:40]}]  AUC: {auc:.4f}")
        df[col] = clf.predict_proba(X)[:,1]

    print("✅ Cross-sell models done")
    return df

df = train_crosssell_model(df)


# ─────────────────────────────────────────────────────────────
# CELL 11 — Lookalike Acquisition
# ─────────────────────────────────────────────────────────────
def build_lookalike_model(df: pd.DataFrame):
    print("\n🔍 Building lookalike acquisition model …")
    LOOK_FEATURES = [
        "Age","Sex_Encoded","Region_Encoded",
        "Log_End_Balance","Account_Tenure_Years",
        "RFM_Score","Balance_Growth_Rate","Product_Encoded",
    ]
    df["Is_High_Value"] = (
        df["CLV_Predicted"] >= df["CLV_Predicted"].quantile(0.90)
    ).astype(int)

    X = df[LOOK_FEATURES].fillna(0)
    y = df["Is_High_Value"]
    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=RANDOM_SEED
    )
    model = RandomForestClassifier(
        n_estimators=300, max_depth=8, class_weight="balanced",
        random_state=RANDOM_SEED, n_jobs=-1
    )
    model.fit(X_tr, y_tr)
    auc = roc_auc_score(y_te, model.predict_proba(X_te)[:,1])
    print(f"   Lookalike AUC: {auc:.4f}")

    df["Lookalike_Score"] = model.predict_proba(X)[:,1]
    top_targets = df.nlargest(1000,"Lookalike_Score")[
        ["CIF","Account Name","Region","Age","Sex",
         "CLV_Predicted","Lookalike_Score","Segment_Label"]
    ]
    top_targets.to_csv(OUTPUT_DIR / "lookalike_targets_real.csv", index=False)
    print("✅ Lookalike model done — top 1,000 targets saved")
    display(top_targets.head(10))   # ← show top 10 in Colab
    return model, df

lookalike_model, df = build_lookalike_model(df)


# ─────────────────────────────────────────────────────────────
# CELL 12 — Export All Outputs + Business Dashboard
# ─────────────────────────────────────────────────────────────
def export_outputs(df: pd.DataFrame, all_metrics: dict):
    print("\n💾 Exporting all outputs …")

    out_cols = [
        "CIF","Account Number","Account Name","Branch Code","Branch Name",
        "Sex","Age","Age_Group","Region","Product Type","Product Code",
        "Opening Date","Last_Transaction_Date",
        "Initial_Deposit","End Balance",
        "Account_Tenure_Years","Days_Since_Last_Txn",
        "ACTIVE_ACCOUNT","INOPERATIVE","DORMANT",
        "Debit Freez","Credit Freez",
        "RFM_Score","RFM_Segment","Segment_Label",
        "CLV_Predicted","CLV_Tier",
        "Churn_Probability","Churn_Risk_Band",
        "Dormancy_Risk_Score","Dormancy_Risk_Band",
        "Lookalike_Score",
    ]
    out_cols = [c for c in out_cols if c in df.columns]
    df[out_cols].to_csv(OUTPUT_DIR / "ahadu_bank_enriched_real.csv", index=False)
    print(f"   ✅ Enriched dataset saved — {len(df):,} rows")

    df[df["DORMANT"] == 1].sort_values("CLV_Predicted", ascending=False)[
        ["CIF","Account Name","Region","Product Type",
         "End Balance","Days_Since_Last_Txn","CLV_Predicted","Churn_Probability"]
    ].to_csv(OUTPUT_DIR / "dormant_reengagement_real.csv", index=False)
    print(f"   ✅ Dormant re-engagement list saved — {df['DORMANT'].sum():,} accounts")

    # Business KPI Dashboard
    fig, axes = plt.subplots(2, 3, figsize=(18,10))
    fig.suptitle("Ahadu Bank  Business Intelligence Dashboard ",
                 fontsize=14, fontweight="bold")

    clv_c = df["CLV_Tier"].value_counts()
    axes[0,0].bar(clv_c.index.astype(str), clv_c.values, color="#4C72B0")
    axes[0,0].set_title("CLV Tier Distribution")

    churn_c = df["Churn_Risk_Band"].value_counts()
    colors  = ["#2ca02c","#ff7f0e","#d62728","#8B0000"]
    axes[0,1].bar(churn_c.index.astype(str), churn_c.values,
                  color=colors[:len(churn_c)])
    axes[0,1].set_title("Churn Risk Distribution")

    seg_c = df["Segment_Label"].value_counts()
    axes[0,2].barh(seg_c.index[::-1], seg_c.values[::-1], color="#DD8452")
    axes[0,2].set_title("Customer Segments")

    clv_seg = df.groupby("Segment_Label")["CLV_Predicted"].mean().sort_values()
    axes[1,0].barh(clv_seg.index, clv_seg.values, color="#55A868")
    axes[1,0].set_title("Avg CLV by Segment")

    dorm_r = df.groupby("Region")["Dormancy_Risk_Score"].mean().sort_values(ascending=False)
    axes[1,1].bar(range(len(dorm_r)), dorm_r.values, color="#C44E52")
    axes[1,1].set_xticks(range(len(dorm_r)))
    axes[1,1].set_xticklabels(dorm_r.index, rotation=45, ha="right", fontsize=8)
    axes[1,1].set_title("Dormancy Risk by Region")

    axes[1,2].hist(df["Churn_Probability"], bins=50, color="#8172B2", edgecolor="white")
    axes[1,2].axvline(0.7, color="red", linestyle="--", label="70% threshold")
    axes[1,2].set_title("Churn Probability Distribution")
    axes[1,2].legend()

    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "business_dashboard_real.png", dpi=120, bbox_inches="tight")
    plt.show()   # ← visible in Colab

    summary = {
        "run_timestamp"            : datetime.now().isoformat(),
        "data_source"              : "real",
        "total_customers"          : int(len(df)),
        "active_accounts"          : int(df["ACTIVE_ACCOUNT"].sum()),
        "inoperative_accounts"     : int(df["INOPERATIVE"].sum()),
        "dormant_accounts"         : int(df["DORMANT"].sum()),
        "avg_clv_predicted"        : float(round(df["CLV_Predicted"].mean(), 2)),
        "high_churn_risk_customers": int((df["Churn_Probability"] > 0.70).sum()),
        "model_metrics"            : all_metrics,
    }
    with open(OUTPUT_DIR / "pipeline_summary_real.json","w") as f:
        json.dump(summary, f, indent=2)

    print("\n" + "="*55)
    print("  PIPELINE COMPLETE ✅")
    print("="*55)
    print(json.dumps(summary, indent=2))
    print(f"\n📁 All files saved to: outputs_real/")
    print("   • ahadu_bank_enriched_real.csv")
    print("   • nbp_recommendations_real.csv")
    print("   • dormant_reengagement_real.csv")
    print("   • dormancy_at_risk_real.csv")
    print("   • lookalike_targets_real.csv")
    print("   • segment_summary_real.csv")
    print("   • eda_dashboard_real.png")
    print("   • business_dashboard_real.png")
    print("   • clv_feature_importance_real.png")
    print("   • churn_confusion_matrix_real.png")
    print("   • pipeline_summary_real.json")

export_outputs(df, {"clv": clv_metrics, "churn": churn_metrics})


# ─────────────────────────────────────────────────────────────
# CELL 13 — Download all output files from Colab
# ─────────────────────────────────────────────────────────────
from google.colab import files

output_files = [
    "outputs_real/ahadu_bank_enriched_real.csv",
    "outputs_real/nbp_recommendations_real.csv",
    "outputs_real/dormant_reengagement_real.csv",
    "outputs_real/dormancy_at_risk_real.csv",
    "outputs_real/lookalike_targets_real.csv",
    "outputs_real/segment_summary_real.csv",
    "outputs_real/pipeline_summary_real.json",
    "outputs_real/eda_dashboard_real.png",
    "outputs_real/business_dashboard_real.png",
    "outputs_real/clv_feature_importance_real.png",
    "outputs_real/churn_confusion_matrix_real.png",
    "outputs_real/elbow_curve_real.png",
    "outputs_real/segmentation_pca_real.png",
]

for f in output_files:
    if Path(f).exists():
        files.download(f)
        print(f"⬇️  Downloaded: {f}")
    else:
        print(f"⚠️  Not found (may not have been generated): {f}")

# ─────────────────────────────────────────────────────────────
# CELL 14 — Gradio Demo App (FIXED)
# ─────────────────────────────────────────────────────────────
import gradio as gr
import joblib
import numpy as np
import pandas as pd

# ── Save models ───────────────────────────────────────────────
joblib.dump(clv_model,      "outputs_real/clv_model.pkl")
joblib.dump(churn_model,    "outputs_real/churn_model.pkl")
joblib.dump(dormancy_model, "outputs_real/dormancy_model.pkl")
print("✅ Models saved")

# ── Shared data (NOT components) ─────────────────────────────
regions  = sorted(df["Region"].dropna().unique().tolist())
products = sorted(df["Product Type"].dropna().unique().tolist())

# ─────────────────────────────────────────────────────────────
# CLV PREDICTION
# ─────────────────────────────────────────────────────────────
def predict_clv(age, sex, region, product,
                initial_deposit, end_balance, tenure_years,
                days_since_txn, rfm_score, balance_growth_rate,
                product_count, dormant, inoperative,
                active, debit_freez, credit_freez):
    sex_enc  = 1 if sex == "Male" else 0
    le_reg   = LabelEncoder().fit(df["Region"])
    le_prod  = LabelEncoder().fit(df["Product Type"])
    reg_enc  = le_reg.transform([region])[0]  if region  in le_reg.classes_ else 0
    prod_enc = le_prod.transform([product])[0] if product in le_prod.classes_ else 0
    log_init    = np.log1p(initial_deposit)
    log_bal     = np.log1p(end_balance)
    b2d         = round(end_balance / (initial_deposit + 1), 4)
    tenure_days = int(tenure_years * 365.25)
    r = f = m   = 3
    input_data  = np.array([[
        age, sex_enc, reg_enc, prod_enc,
        log_init, log_bal, tenure_years,
        days_since_txn, rfm_score, r, f, m,
        balance_growth_rate, b2d, product_count,
        tenure_days, dormant, inoperative, active,
        debit_freez, credit_freez
    ]])
    clv_val = clv_model.predict(input_data)[0]
    tier    = "🥇 Top-CLV"  if clv_val > df["CLV_Predicted"].quantile(0.75) else \
              "🥈 High-CLV" if clv_val > df["CLV_Predicted"].quantile(0.50) else \
              "🥉 Mid-CLV"  if clv_val > df["CLV_Predicted"].quantile(0.25) else \
              "⬇️  Low-CLV"
    return f"ETB {clv_val:,.2f}", tier

clv_tab = gr.Interface(
    fn=predict_clv,
    inputs=[
        gr.Number(label="Age",                        value=35),
        gr.Radio(["Male","Female"], label="Sex",      value="Male"),
        gr.Dropdown(regions,  label="Region"),
        gr.Dropdown(products, label="Product Type"),
        gr.Number(label="Initial Deposit (ETB)",      value=5000),
        gr.Number(label="End Balance (ETB)",          value=12000),
        gr.Number(label="Account Tenure (Years)",     value=2.5),
        gr.Number(label="Days Since Last Transaction",value=45),
        gr.Number(label="RFM Score (3–15)",           value=9),
        gr.Number(label="Balance Growth Rate",        value=0.4),
        gr.Number(label="Product Count",              value=1),
        gr.Radio([0,1], label="Dormant (0=No 1=Yes)",    value=0),
        gr.Radio([0,1], label="Inoperative (0=No 1=Yes)",value=0),
        gr.Radio([0,1], label="Active (0=No 1=Yes)",     value=1),
        gr.Radio([0,1], label="Debit Freeze",            value=0),
        gr.Radio([0,1], label="Credit Freeze",           value=0),
    ],
    outputs=[
        gr.Textbox(label="💰 Predicted CLV"),
        gr.Textbox(label="🏷️ CLV Tier"),
    ],
    title="💰 Customer Lifetime Value Predictor",
    description="Predict the lifetime value of an Ahadu Bank customer."
)

# ─────────────────────────────────────────────────────────────
# CHURN PREDICTION
# ─────────────────────────────────────────────────────────────
def predict_churn(age, sex, region, product,
                  initial_deposit, end_balance, tenure_years,
                  days_since_txn, rfm_score, balance_growth_rate,
                  clv_predicted, debit_freez, credit_freez,
                  product_count):
    sex_enc  = 1 if sex == "Male" else 0
    le_reg   = LabelEncoder().fit(df["Region"])
    le_prod  = LabelEncoder().fit(df["Product Type"])
    reg_enc  = le_reg.transform([region])[0]  if region  in le_reg.classes_ else 0
    prod_enc = le_prod.transform([product])[0] if product in le_prod.classes_ else 0
    log_bal  = np.log1p(end_balance)
    log_init = np.log1p(initial_deposit)
    b2d      = round(end_balance / (initial_deposit + 1), 4)
    r = f = m = 3
    input_data = np.array([[
        age, sex_enc, reg_enc, prod_enc,
        log_bal, log_init, tenure_years,
        days_since_txn, rfm_score, r, f, m,
        balance_growth_rate, b2d, clv_predicted,
        debit_freez, credit_freez, product_count
    ]])
    churn_prob = churn_model.predict_proba(input_data)[0][1]
    risk_band  = "🔴 Critical Risk" if churn_prob > 0.70 else \
                 "🟠 High Risk"     if churn_prob > 0.50 else \
                 "🟡 Medium Risk"   if churn_prob > 0.30 else \
                 "🟢 Low Risk"
    return f"{churn_prob*100:.1f}%", risk_band

churn_tab = gr.Interface(
    fn=predict_churn,
    inputs=[
        gr.Number(label="Age",                        value=35),
        gr.Radio(["Male","Female"], label="Sex",      value="Male"),
        gr.Dropdown(regions,  label="Region"),
        gr.Dropdown(products, label="Product Type"),
        gr.Number(label="Initial Deposit (ETB)",      value=5000),
        gr.Number(label="End Balance (ETB)",          value=12000),
        gr.Number(label="Account Tenure (Years)",     value=2.5),
        gr.Number(label="Days Since Last Transaction",value=45),
        gr.Number(label="RFM Score (3–15)",           value=9),
        gr.Number(label="Balance Growth Rate",        value=0.4),
        gr.Number(label="CLV Predicted (ETB)",        value=50000),
        gr.Radio([0,1], label="Debit Freeze",         value=0),
        gr.Radio([0,1], label="Credit Freeze",        value=0),
        gr.Number(label="Product Count",              value=1),
    ],
    outputs=[
        gr.Textbox(label="⚠️ Churn Probability"),
        gr.Textbox(label="🚦 Risk Band"),
    ],
    title="⚠️ Churn Risk Predictor",
    description="Predict whether a customer is at risk of going dormant."
)

# ─────────────────────────────────────────────────────────────
# DORMANCY DETECTION
# ─────────────────────────────────────────────────────────────
def predict_dormancy(age, sex, region, product,
                     end_balance, tenure_years, days_since_txn,
                     balance_growth_rate, rfm_score,
                     churn_probability, debit_freez, credit_freez):
    sex_enc  = 1 if sex == "Male" else 0
    le_reg   = LabelEncoder().fit(df["Region"])
    le_prod  = LabelEncoder().fit(df["Product Type"])
    reg_enc  = le_reg.transform([region])[0]  if region  in le_reg.classes_ else 0
    prod_enc = le_prod.transform([product])[0] if product in le_prod.classes_ else 0
    log_bal  = np.log1p(end_balance)
    input_data = np.array([[
        age, sex_enc, reg_enc, prod_enc,
        log_bal, tenure_years, days_since_txn,
        balance_growth_rate, rfm_score, churn_probability,
        debit_freez, credit_freez
    ]])
    dorm_score = dormancy_model.predict_proba(input_data)[0][1]
    risk_band  = "🔴 High Dormancy Risk" if dorm_score > 0.60 else \
                 "🟡 At Risk"            if dorm_score > 0.30 else \
                 "🟢 Safe"
    return f"{dorm_score*100:.1f}%", risk_band

dormancy_tab = gr.Interface(
    fn=predict_dormancy,
    inputs=[
        gr.Number(label="Age",                        value=35),
        gr.Radio(["Male","Female"], label="Sex",      value="Male"),
        gr.Dropdown(regions,  label="Region"),
        gr.Dropdown(products, label="Product Type"),
        gr.Number(label="End Balance (ETB)",          value=12000),
        gr.Number(label="Account Tenure (Years)",     value=2.5),
        gr.Number(label="Days Since Last Transaction",value=45),
        gr.Number(label="Balance Growth Rate",        value=0.4),
        gr.Number(label="RFM Score (3–15)",           value=9),
        gr.Number(label="Churn Probability (0–1)",    value=0.2),
        gr.Radio([0,1], label="Debit Freeze",         value=0),
        gr.Radio([0,1], label="Credit Freeze",        value=0),
    ],
    outputs=[
        gr.Textbox(label="😴 Dormancy Risk Score"),
        gr.Textbox(label="🏷️ Risk Band"),
    ],
    title="😴 Dormancy Risk Detector",
    description="Detect accounts at risk of becoming inoperative."
)

# ─────────────────────────────────────────────────────────────
# NBP RECOMMENDER
# ─────────────────────────────────────────────────────────────
def recommend_nbp(age, sex, region, product,
                  end_balance, tenure_years,
                  days_since_txn, rfm_score,
                  clv_predicted, churn_probability):
    sex_enc   = 1 if sex == "Male" else 0
    rfm_segment = "Champions"  if rfm_score >= 12 else \
                  "High Value" if rfm_score >= 9  else \
                  "Mid Value"  if rfm_score >= 6  else \
                  "Low Value"
    clv_tier  = "Top-CLV"  if clv_predicted >= df["CLV_Predicted"].quantile(0.75) else \
                "High-CLV" if clv_predicted >= df["CLV_Predicted"].quantile(0.50) else \
                "Mid-CLV"  if clv_predicted >= df["CLV_Predicted"].quantile(0.25) else \
                "Low-CLV"
    age_group = "18-25" if age <= 25 else \
                "26-35" if age <= 35 else \
                "36-45" if age <= 45 else \
                "46-55" if age <= 55 else "56+"
    churn_band = "Critical Risk" if churn_probability > 0.70 else \
                 "High Risk"     if churn_probability > 0.50 else \
                 "Medium Risk"   if churn_probability > 0.30 else \
                 "Low Risk"
    seg_product = pd.crosstab(df["Segment_Label"], df["Product Type"], normalize="index")
    clv_product = pd.crosstab(df["CLV_Tier"],      df["Product Type"], normalize="index")
    age_product = pd.crosstab(df["Age_Group"],     df["Product Type"], normalize="index")
    scores = {}
    for prod in PRODUCT_LIST:
        if prod == product: continue
        score = 0.0
        if rfm_segment in seg_product.index and prod in seg_product.columns:
            score += seg_product.loc[rfm_segment, prod] * 0.40
        if clv_tier in clv_product.index and prod in clv_product.columns:
            score += clv_product.loc[clv_tier, prod] * 0.35
        if age_group in age_product.index and prod in age_product.columns:
            score += age_product.loc[age_group, prod] * 0.25
        scores[prod] = round(score, 5)
    top3   = sorted(scores.items(), key=lambda x: x[1], reverse=True)[:3]
    medals = ["🥇","🥈","🥉"]
    recs   = "\n\n".join([
        f"{medals[i]}  Rank {i+1}: {p}\n     Confidence: {c*100:.2f}%  {'█'*int(c*200)}"
        for i, (p,c) in enumerate(top3)
    ])
    profile = (f"👤 Customer Profile\n{'─'*38}\n"
               f"  Age Group  : {age_group}\n  Region     : {region}\n"
               f"  Sex        : {sex}\n  RFM Segment: {rfm_segment}\n"
               f"  CLV Tier   : {clv_tier}\n  Churn Risk : {churn_band}\n"
               f"  Current    : {product}\n{'─'*38}")
    if churn_probability > 0.50:
        action = f"⚠️  High churn risk! Prioritize retention\n    offer with {top3[0][0]}."
    elif clv_tier in ["Top-CLV","High-CLV"]:
        action = f"🌟 High-value customer — offer premium\n    product: {top3[0][0]}."
    else:
        action = f"📈 Cross-sell opportunity:\n    {top3[0][0]}."
    return profile, recs, action

nbp_tab = gr.Interface(
    fn=recommend_nbp,
    inputs=[
        gr.Number(label="Age",                        value=35),
        gr.Radio(["Male","Female"], label="Sex",      value="Male"),
        gr.Dropdown(regions,  label="Region"),
        gr.Dropdown(products, label="Current Product"),
        gr.Number(label="End Balance (ETB)",          value=12000),
        gr.Number(label="Account Tenure (Years)",     value=2.5),
        gr.Number(label="Days Since Last Transaction",value=45),
        gr.Number(label="RFM Score (3–15)",           value=9),
        gr.Number(label="CLV Predicted (ETB)",        value=50000),
        gr.Slider(minimum=0.0, maximum=1.0, step=0.01,
                  label="Churn Probability",          value=0.2),
    ],
    outputs=[
        gr.Textbox(label="👤 Customer Profile",  lines=10),
        gr.Textbox(label="🎯 Recommendations",   lines=10),
        gr.Textbox(label="💡 Action Insight",    lines=4),
    ],
    title="🎯 Next Best Product (NBP) Recommender",
    description="Top 3 personalized product recommendations."
)

# ─────────────────────────────────────────────────────────────
# DORMANT RE-ENGAGEMENT
# ─────────────────────────────────────────────────────────────
def dormant_reengagement(region, product, min_balance,
                          max_days_inactive, top_n,
                          sort_by, risk_filter):
    dormant_df = df[df["DORMANT"] == 1].copy()
    if region  != "All Regions":  dormant_df = dormant_df[dormant_df["Region"]      == region]
    if product != "All Products": dormant_df = dormant_df[dormant_df["Product Type"] == product]
    dormant_df = dormant_df[dormant_df["End Balance"]          >= min_balance]
    dormant_df = dormant_df[dormant_df["Days_Since_Last_Txn"]  <= max_days_inactive]
    if risk_filter != "All Risk Levels":
        dormant_df = dormant_df[dormant_df["Churn_Risk_Band"] == risk_filter]
    sort_col_map = {
        "CLV (Highest First)"          : ("CLV_Predicted",       False),
        "Balance (Highest First)"      : ("End Balance",         False),
        "Dormancy Risk (Highest First)": ("Dormancy_Risk_Score", False),
        "Days Inactive (Most First)"   : ("Days_Since_Last_Txn", False),
    }
    col, asc   = sort_col_map.get(sort_by, ("CLV_Predicted", False))
    dormant_df = dormant_df.sort_values(col, ascending=asc)
    top_df     = dormant_df.head(int(top_n))

    total_found    = len(dormant_df)
    total_balance  = dormant_df["End Balance"].sum()
    avg_clv        = dormant_df["CLV_Predicted"].mean()
    avg_inactive   = dormant_df["Days_Since_Last_Txn"].mean()
    critical_count = (dormant_df["Churn_Risk_Band"] == "Critical Risk").sum() \
                     if "Churn_Risk_Band" in dormant_df.columns else 0

    summary = (f"📊 Re-engagement Summary\n{'─'*43}\n"
               f"  Total dormant accounts : {total_found:,}\n"
               f"  Balance at risk        : ETB {total_balance:,.2f}\n"
               f"  Average CLV            : ETB {avg_clv:,.2f}\n"
               f"  Avg days inactive      : {avg_inactive:,.0f} days\n"
               f"  Critical risk accounts : {critical_count:,}\n{'─'*43}\n"
               f"  Showing top {int(top_n)} sorted by: {sort_by}")

    display_cols = {
        "Account Name"       : "Name",
        "Region"             : "Region",
        "Product Type"       : "Product",
        "End Balance"        : "Balance (ETB)",
        "Days_Since_Last_Txn": "Days Inactive",
        "CLV_Predicted"      : "CLV (ETB)",
        "Dormancy_Risk_Score": "Dormancy %",
        "Churn_Risk_Band"    : "Churn Risk",
    }
    available  = {k: v for k, v in display_cols.items() if k in top_df.columns}
    result_df  = top_df[list(available.keys())].copy()
    result_df.columns = list(available.values())
    if "Balance (ETB)" in result_df.columns:
        result_df["Balance (ETB)"] = result_df["Balance (ETB)"].apply(lambda x: f"ETB {x:,.2f}")
    if "CLV (ETB)" in result_df.columns:
        result_df["CLV (ETB)"]     = result_df["CLV (ETB)"].apply(lambda x: f"ETB {x:,.2f}")
    if "Dormancy %" in result_df.columns:
        result_df["Dormancy %"]    = result_df["Dormancy %"].apply(lambda x: f"{x*100:.1f}%")

    if total_found == 0:
        action = "✅ No dormant accounts match your filters."
    elif avg_clv > df["CLV_Predicted"].quantile(0.75):
        action = ("🌟 HIGH-VALUE segment!\n"
                  "  → Assign relationship manager\n"
                  "  → Personal call within 48 hours\n"
                  f"  → Offer premium product upgrade\n"
                  f"  💰 Recoverable: ETB {total_balance:,.2f}")
    elif avg_clv > df["CLV_Predicted"].quantile(0.50):
        action = ("📈 MID-VALUE segment.\n"
                  "  → Personalized SMS/email campaign\n"
                  "  → Offer fee waiver or bonus interest\n"
                  f"  💰 Recoverable: ETB {total_balance:,.2f}")
    else:
        action = ("📩 LOW-VALUE segment.\n"
                  "  → Automated reactivation campaign\n"
                  "  → Low-barrier re-engagement incentive\n"
                  f"  💰 Recoverable: ETB {total_balance:,.2f}")
    if critical_count > 0:
        action += f"\n\n  ⚠️  {critical_count:,} CRITICAL accounts need immediate action!"

    return summary, result_df, action

reengagement_tab = gr.Interface(
    fn=dormant_reengagement,
    inputs=[
        gr.Dropdown(["All Regions"]  + regions,  label="Filter by Region",  value="All Regions"),
        gr.Dropdown(["All Products"] + products, label="Filter by Product", value="All Products"),
        gr.Number(label="Minimum Balance (ETB)", value=0),
        gr.Slider(minimum=30,  maximum=3650, step=30,  label="Max Days Inactive",          value=1825),
        gr.Slider(minimum=5,   maximum=100,  step=5,   label="Number of Accounts to Show", value=20),
        gr.Radio([
            "CLV (Highest First)",
            "Balance (Highest First)",
            "Dormancy Risk (Highest First)",
            "Days Inactive (Most First)",
        ], label="Sort By", value="CLV (Highest First)"),
        gr.Dropdown([
            "All Risk Levels","Critical Risk",
            "High Risk","Medium Risk","Low Risk"
        ], label="Filter by Churn Risk", value="All Risk Levels"),
    ],
    outputs=[
        gr.Textbox(label="📊 Summary",           lines=9),
        gr.Dataframe(label="🏦 Dormant Accounts", wrap=True),
        gr.Textbox(label="📋 Action Plan",        lines=8),
    ],
    title="🔄 Dormant Account Re-engagement",
    description="Identify and prioritize dormant accounts for re-engagement campaigns."
)

# ─────────────────────────────────────────────────────────────
# LAUNCH — ALL 5 TABS
# ─────────────────────────────────────────────────────────────
app = gr.TabbedInterface(
    interface_list=[clv_tab, churn_tab, dormancy_tab, nbp_tab, reengagement_tab],
    tab_names=[
        "💰 CLV Prediction",
        "⚠️  Churn Risk",
        "😴 Dormancy Risk",
        "🎯 NBP Recommender",
        "🔄 Re-engagement",
    ]
)

print("\n🚀 Launching Ahadu Bank AI Platform — 5 Tabs ...")
app.launch(share=True)
```
